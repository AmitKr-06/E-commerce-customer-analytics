# 01 — Business Understanding
**Project:** E-Commerce Customer Analytics & Revenue Intelligence
**Dataset:** Olist Brazilian E-Commerce (Kaggle), ~100k orders, 2016–2018

## 1. Business context
Olist is a Brazilian marketplace connecting small sellers to customers through a single platform. 
The company holds large volumes of transaction data but has limited insight into who its 
most valuable customers are, where revenue comes from, and how delivery performance 
affects customer satisfaction.

## 2. Problem statement
Olist cannot clearly identify its high-value customer segments, explain revenue trends, or 
determine how delivery delays affect customer experience. This limits its ability to target 
marketing, improve seller operations, and retain customers.

## 3. Project objective
Analyze transaction data to (1) measure revenue performance, (2) understand customer 
purchasing behaviour, (3) segment customers by value, (4) quantify the relationship between 
delivery performance and reviews, (5) predict late deliveries, and (6) recommend actions 
supported by evidence.

## 5. Key definitions (used consistently in all notebooks)
- **Revenue:** sum of `payment_value` per order (order_payments). Cross-checked against price + freight_value in notebook 4.
- **Late delivery:** `order_delivered_customer_date` > `order_estimated_delivery_date`.
- **Delivery delay (days):** actual delivery date − estimated delivery date.
- **Customer:** identified by `customer_unique_id` (NOT `customer_id`, which is created per order).
- **Repeat customer:** a `customer_unique_id` with more than one distinct order.
- **Analysis population:** orders with status `delivered` for delivery and review analysis.

## 6. Success metrics (KPIs)
Total revenue · Monthly revenue growth · Average order value · Repeat purchase rate · 
Late delivery rate · Average review score · Revenue share of top customer segment · 
Late-delivery model Recall / F1 / ROC-AUC

## 7. Scope, assumptions and limitations
- Data covers 2016–2018 only; trends may not reflect today.
- No customer demographics or marketing data, so segmentation relies on behaviour.
- Most customers purchase once, so repeat-purchase analysis will be limited.
- Review scores are subjective and only exist for customers who chose to review.
- Findings show association, not causation.
